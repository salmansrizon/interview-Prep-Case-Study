# 01 — Exploratory Data Analysis
## Customer Segmentation Engine

Use this notebook to explore the synthetic dataset before clustering.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

# Load synthetic data
df = pd.read_csv('../data/processed/synthetic_customers.csv')
df.head()
```

```python
# Distribution of customer segments
df['segment_ground_truth'].value_counts().plot(kind='bar', title='Ground Truth Segments')
plt.show()
```

```python
# Correlation heatmap
numeric_cols = df.select_dtypes(include=['float64', 'int64']).columns
sns.heatmap(df[numeric_cols].corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Feature Correlations')
plt.show()
```

```python
# Pairplot of key features
sns.pairplot(df, hue='segment_ground_truth', 
             vars=['purchase_frequency', 'avg_order_value', 'total_spend', 'satisfaction_score'],
             palette='Set2')
plt.show()
```
