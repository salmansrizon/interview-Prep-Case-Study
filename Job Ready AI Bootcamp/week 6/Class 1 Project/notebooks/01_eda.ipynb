# 01 - Exploratory Data Analysis

Use this notebook to explore the dataset before training.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

# Load synthetic or uploaded data
df = pd.read_csv('../data/processed/synthetic_spam.csv')
df.head()
```

```python
# Class distribution
df['label'].value_counts().plot(kind='bar', title='Class Distribution')
plt.show()
```

```python
# Text length analysis
df['length'] = df['text'].str.len()
sns.boxplot(data=df, x='label', y='length')
plt.show()
```
