2. Run করার আগে expected output কী হতে পারে লিখে ভাবুন।

5. Notebook শেষে নিজের ভাষায় লিখুন: এটি কোন problem solve করেছে, কীভাবে করেছে এবং limitation কী।

# 01 — Neural Network Exploration
## Understanding Forward Pass, Backpropagation, and Weight Updates

Use this notebook to step through each operation manually.

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent))

import numpy as np
import matplotlib.pyplot as plt
from src.core.neuron import Neuron
from src.core.activation import ActivationFunctions
from src.core.network import NeuralNetwork
from src.gates.logic_gates import LogicGateDataset
from src.visualization.nn_viz import NetworkVisualizer
```

```python
# Step 1: Create a single neuron and trace its computation
neuron = Neuron(weights=[0.5, -0.3], bias=0.1)
inputs = [1.0, 2.0]

z = neuron.compute_z(inputs)
print(f'Weighted sum z = {z:.4f}')

# Try with different activations
act = ActivationFunctions()
print(f'Sigmoid(z) = {act.sigmoid(np.array([z]))[0]:.4f}')
print(f'ReLU(z) = {act.relu(np.array([z]))[0]:.4f}')
```

```python
# Step 2: Build a 2-layer network and inspect weights
nn = NeuralNetwork(input_size=2, hidden_size=3, output_size=1)

print('W1 shape:', nn.W1.shape)
print('W1 values:')
print(nn.W1.round(3))
print('\nW2 shape:', nn.W2.shape)
print('W2 values:')
print(nn.W2.round(3))
```

```python
# Step 3: Manual forward pass for XOR input [1, 0]
X = np.array([[1, 0]])
y = np.array([[1]])  # XOR(1,0) = 1

# Forward pass
output = nn.forward(X)
print(f'Prediction before training: {output[0,0]:.4f}')
print(f'Target: {y[0,0]}')
print(f'Initial loss: {nn.compute_loss(y, output):.4f}')
```

```python
# Step 4: Train and observe weight changes
dataset = LogicGateDataset('XOR')
X, y = dataset.get_data()

print('W1 BEFORE training:')
print(nn.W1.round(3))

history = nn.train(X, y, epochs=1000, verbose=False)

print('\nW1 AFTER training:')
print(nn.W1.round(3))

print(f'\nLoss before: {history["loss"][0]:.4f}')
print(f'Loss after: {history["loss"][-1]:.4f}')
```

```python
# Step 5: Visualize decision boundary
viz = NetworkVisualizer()
fig = viz.plot_decision_boundary(nn, X, y, title='XOR')
plt.show()
```
